
### Task 3.1
Write code to load the triples created in Task 1.2 into a knowledge graph using the [RDFLib](https://rdflib.readthedocs.io/en/stable/) Python library. Print the triples and the number of triples contained in the graph.

```python
# Code Here
G = Graph()
G.parse("triples.ttl")
nm = G.namespace_manager
for s, p, o in G:
    print(f"{s.n3(nm)} -> {p.n3(nm)} -> {o.n3(nm)}")
```

```python
print("Number of Triples : {}".format(len(G)))
```

### Task 3.2
Write code to show all cities in your graph.

```python
DBO = Namespace("https://dbpedia.org/ontology/")

for city in G.subjects(RDF.type, DBO.City):
    print(city)
```

```python
q = """
PREFIX dbo: <https://dbpedia.org/ontology/>
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>

SELECT ?city ?label WHERE {
    ?city a dbo:City .
}
"""

for row in G.query(q):
    print(row.city)
```

Run the SPARQL query from Task 2.3 on your graph.

```python
query = """
SELECT ?inhabitants WHERE {
    dbr:Germany dbo:capital ?capital .
    ?capital dbo:populationTotal ?inhabitants .
}
"""
# Code Here
result = G.query(query)
for row in result:
    print(f"Number of Inhabitants : {row.inhabitants}")
```
